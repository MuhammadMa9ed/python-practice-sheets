Task 1 Process a List

```python
scores = [75, 90, 60, 85, 90, 50, 75, 95]
```

```python
# 1. Create a list of unique scores without changing the original list. Sort it in descending order
Unique_scores = sorted(set(scores), reverse=True)
print(Unique_scores)
```

```python
# 2. Create a list of scores greater than or equal to 70, preserving duplicates.
Filtered_list = []
for score in scores:
    if score >= 70:
        Filtered_list.append(score)
print(Filtered_list)
```

```python
# 3. Find the second-highest distinct score and calculate the average of the original scores.
Second_highest_score = Unique_scores[1]
Average_score = sum(scores) / len(scores)
print(Second_highest_score)
print(Average_score)
```

```python
# 4. Create a new list where scores below 60 are replaced with 60; keep other scores unchanged.
New_list = []
for score in scores:
    if score < 60:
        New_list.append(60)
    else:
        New_list.append(score)
print(New_list)
```

```python
# 5. Create a dictionary that stores each distinct score as a key and its frequency as the value.
Score_dict = {}
for score in scores:
    if score in Score_dict:
        Score_dict[score] += 1
    else:
        Score_dict[score] = 1
print(Score_dict)
```

Task 4 Build Reusable Functions

```python
# 1. Write calculate_average(scores): return the average, or 0 for an empty list
def calculate_average(scores):
    if len(scores) == 0:
        return 0
    average_score = sum(scores)/len(scores)
    return average_score
```

```python
calculate_average(scores)
```

```python
# 2. Write get_result(score, passing_score=60): return "Pass" or "Fail".
def get_result(score, passing_score=60):
    if score >= passing_score:
        return "Pass"
    else:
        return "Fail"
```

```python
''' 3. Write summarize_student(name, scores): return a dictionary with name, average_score, and result.  
    Use the two functions above; use the unrounded average to determine the result.
'''
def summarize_student(name, scores):
    Average_score = calculate_average(scores)
    result = get_result(Average_score)
    return {
        "name": name,
        "average_score": Average_score,
        "result":result
    }
```

```python
# 4. Test an empty list, an average exactly equal to 60, and a custom passing score of 70.
print("Test 1 (Empty):", summarize_student("Ali", []))
print("Test 2 (Exact 60):", summarize_student("Sara",[60,60]))
avg_omar = calculate_average([60,60,10])  # المتوسط هيكون 65.0
print("Test 3 (Custom 70):", get_result(avg_omar, passing_score=70))
```

```python
# 5. Call summarize_student using positional arguments and then keyword arguments.
# positional arguments
summarize_student("Mona", [60,60,55])
```

```python
# keyword arguments
summarize_student(scores=[60,60,55], name="Mona")
```

```python
''' 
    6. Write count_results(summaries): return a dictionary containing the number of passing and failing students.
    Test it with an empty list as well.
'''
def count_results(summaries):
    counts = {"Pass": 0, "Fail": 0}
    
    for student in summaries:
        if student["result"] == "Pass":
            counts["Pass"] += 1
        elif student["result"] == "Fail":
            counts["Fail"] += 1  
    return counts

```

Task 7 Clean Customer Records

```python
customers = [
 {"id": 101, "name": " Ahmed ", "city": "cairo"},
 {"id": 102, "name": "Sara", "city": "GIZA"},
 {"id": 101, "name": "Ahmed", "city": "Cairo"},
 {"id": 103, "name": " Omar", "city": "cairo"},
 {"id": 104, "name": "Mona ", "city": "giza"}
]
```

```python
# 1. Write clean_customers(records) to return a new list of cleaned records. Remove extra spaces from names and convert city names to title case. Do not change the input list.
def clean_customers(records):
    cleaned_records = []
    for record in records:
        cleaned_record = {
            "id": record["id"],
            "name": record["name"].strip(),
            "city": record["city"].title()
        }
        cleaned_records.append(cleaned_record)
    return cleaned_records
```

```python
# 2. Remove duplicate customer IDs, keeping the first occurrence of each ID
list_without_duplicates = []
seen_ids = set()
for record in customers:
    if record["id"] in seen_ids:
        continue
    else:
        seen_ids.add(record["id"])
        list_without_duplicates.append(record)
print(list_without_duplicates)
```

```python
# 3. Create a dictionary mapping each city to the number of unique customers in it.
city_counts = {}

for record in list_without_duplicates:
    city = record["city"]
    if city in city_counts:
        city_counts[city] += 1
    else:
        city_counts[city] = 1

print(city_counts)
```

```python
# 4. Return an empty list when the input is empty. Display the original and cleaned data.
print("Empty list result:", clean_customers([]))
print("Original Data:", customers)
cleaned_data = clean_customers(customers)
print("Cleaned Data:", cleaned_data)
```

Task 8 Summarize Orders

```python
orders = [
 {"customer": "Ahmed", "amount": 1200, "status": "Completed"},
 {"customer": "Sara", "amount": 850, "status": "Completed"},
 {"customer": "Ahmed", "amount": 400, "status": "Cancelled"},
 {"customer": "Omar", "amount": 600, "status": "Completed"},
 {"customer": "Sara", "amount": 300, "status": "Completed"}
]
```

```python
# 1. Write summarize_orders(orders) to calculate total completed order amounts per customer. Exclude cancelled orders.
def summarize_orders(orders):
    totals = {}
    
    for order in orders:
        if order["status"] == "Completed":
            customer = order["customer"]
            amount = order["amount"]
            totals[customer] = totals.get(customer, 0) + amount
            
    return totals
```

```python
# 2. Return the totals as a dictionary. Return an empty dictionary for an empty input list.
customer_totals = summarize_orders(orders)
print("Customer Totals:", customer_totals)
empty_test = summarize_orders([])
print("Empty List Test:", empty_test)
```

```python
# 3. Sort the customer totals in descending order using a lambda expression.
customer_totals = summarize_orders(orders)
sorted_totals = dict(sorted(customer_totals.items(), key=lambda item: item[1], reverse=True))
print(sorted_totals)
```

```python
# 4. Print the customer with the highest completed order total and the overall completed revenue.
if customer_totals:
    highest_customer = max(customer_totals, key=customer_totals.get)
    highest_total = customer_totals[highest_customer]
    overall_revenue = sum(customer_totals.values())
    
    print(f"Customer with highest completed order total: {highest_customer} - Total: {highest_total}")
    print(f"Overall completed revenue: {overall_revenue}")
```

```python
# 5. Write get_customers_above(totals, minimum=1000) to return a sorted list of names whose totals are strictly greater than minimum. Test the default and a custom minimum.
def get_customers_above(totals, minimum=1000):
    above_minimum = []
    for customer, total in totals.items():
        if total > minimum:
            above_minimum.append(customer)
    return sorted(above_minimum)
```
